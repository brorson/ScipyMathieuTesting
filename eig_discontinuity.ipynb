```python
import numpy as np
from scipy.special import mathieu_a, mathieu_b
from scipy.integrate import trapezoid
from numpy import cos, sin
from numpy.linalg import norm
```

```python
Q = np.linspace(0, 50, 5000)
M = np.linspace(50, 60, 10)
error_a = []
error_b = []

def disc_eig_a():
    for m in M:
        error_M = []
        for i in range(len(Q) - 1):
            
            error_m = np.abs(mathieu_a(m, Q[i + 1]) - mathieu_a(m, Q[i]))
            error_M.append(error_m)
            # if np.isclose(m, 1):
            #     print(m, Q[i+1], mathieu_a(m, Q[i + 1]), mathieu_a(m, Q[i]))
        error_a.append(error_M)

def disc_eig_b():
    for m in M:
        error_M = []
        for i in range(len(Q) - 1):
            
            error_m = np.abs(mathieu_b(m, Q[i + 1]) - mathieu_b(m, Q[i]))
            error_M.append(error_m)
            # if np.isclose(m, 1):
            #     print(m, Q[i+1], mathieu_a(m, Q[i + 1]), mathieu_a(m, Q[i]))
        error_b.append(error_M)

disc_eig_a()
disc_eig_b()
```
