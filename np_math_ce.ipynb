```python
# numpy 

def tridiagonal_matrix_np(n, a, b, c):
    M = np.zeros((n, n))
    # print(f'shape a = {a.shape}, shape b = {b.shape}, c shape = {c.shape}')
    for i in range(n):
        M[i][i] = b[i]
        if i > 0:
            M[i][i-1] = a[i]
        if i < n - 1:
            M[i][i+1] = c[i]

    return M
```

```python
def extract_single_column(Sraw, col_idx):
    """Extract a single column from mpmath matrix"""
    rows = Sraw.rows
    S = mp.matrix(rows, 1)
    for i in range(rows):
        S[i, 0] = Sraw[i, col_idx]
    return S

def extract_columns(Sraw, idx):
    """Extract multiple columns from mpmath matrix using list of indices"""
    rows = Sraw.rows
    if isinstance(idx, int):
        # Single column case
        return extract_single_column(Sraw, idx)
    
    # Multiple columns case
    cols = len(idx)
    S = mp.matrix(rows, cols)
    for i in range(rows):
        for j, col_idx in enumerate(idx):
            S[i, j] = Sraw[i, col_idx]
    return S
```

```python
def mpmath_to_numpy_high_precision(S):
    """Convert mpmath matrix to numpy array with extended precision"""
    rows, cols = S.rows, S.cols
    # Use numpy's longdouble for extended precision
    numpy_array = np.array([[np.longdouble(str(S[i,j])) for j in range(cols)] for i in range(rows)])
    return numpy_array
```

```python
def mpmath_slice_even_2Ne_explicit(S, Ne):
    """More explicit version showing the logic"""
    rows = S.rows
    cols_needed = []
    
    # Generate column indices: start=0, stop=2*Ne, step=2
    col = 0
    while col < 2*Ne and col < S.cols:
        cols_needed.append(col)
        col += 2
    
    # Create new matrix
    if not cols_needed:
        return mp.matrix(rows, 0)
    
    ce = mp.matrix(rows, len(cols_needed))
    for i in range(rows):
        for j, col_idx in enumerate(cols_needed):
            ce[i, j] = S[i, col_idx]
    
    return ce
```

```python
def mpmath_column_sign_fix(ce, zidx, Ne):
    """
    mpmath equivalent of:
    for j in range(Ne):
        if ce[zidx, j] < 0:
            ce[:, j] = - ce[:, j]
    
    Args:
        ce: mpmath matrix
        zidx: row index to check for sign
        Ne: number of columns to process
    """
    for j in range(Ne):
        if ce[zidx, j] < 0:
            # Flip the sign of entire column j
            for i in range(ce.rows):
                ce[i, j] = -ce[i, j]
```

```python
def normalize_mpmath_matrix_explicit(ce, h):
    """
    More explicit version ensuring all operations use mpmath
    """
    norm_cols = []
    
    # Convert h to mpmath if it isn't already
    h_mp = mp.mpf(h)
    
    # First loop: calculate normalization factors
    for j in range(ce.cols):
        int_col = mp.mpf(0)
        for i in range(1, ce.rows):
            # Ensure all operations are in mpmath
            term1 = ce[i - 1, j] ** 2
            term2 = ce[i, j] ** 2
            int_f_ij = mp.mpf(0.5) * (term1 + term2) * h_mp
            int_col += int_f_ij
        traps_norm = mp.sqrt(int_col / mp.pi) 
        norm_cols.append(traps_norm)
    
    # Second loop: normalize each column
    for l in range(ce.cols):
        norm_factor = norm_cols[l]
        for i in range(ce.rows):
            ce[i, l] = ce[i, l] / norm_factor
            
```

# Five point Stencil Formula

Here give a formulation for using the collocation method for higher order difference methods. In particular, we will use a five point stencil.

$$f''(x)     \approx \frac{-f(x+2 h)+16 f(x+h)-30 f(x) + 16 f(x-h) - f(x-2h)}{12 h^2}$$

Note: This approximation is $O(h^4)$

We must solve the Mathieu function ODE:
$$ \frac{d^2 S}{dv^2} + (a - 2q \cdot \cos(2v))S = 0$$

We can plug in the stencil as an approximation to the second derivative. To obtain a linear equation in $5$ variables of the following form:
$$ M_2 S_{n+2} + M_1 S_{n+1} + M_{0} S_{n} + M_{-1} S_{n-1} + M_{-2} S_{n-2} + (a -2q \cos(2v_n))S_n = 0$$

We can rearrange this equation to obtain an eigenvalue equation of the form:
$$ MS = a \cdot S$$
where $a$ is the scalar given by the ODE and $M$ is an $n$ by $n$ matrix

It is important to note that this formulation needs a description of the boundary constraints.

For _ce_ functions, the derivative must vanish at $v= \pm \pi$, thus, we use another finite difference scheme, in particular, the central difference scheme for first derivative:
$$f'(x) = \frac{f(x + h) - f(x-h)}{h}$$
If our boundary lies at point $v_2$, then $f'(v_2) = 0$ translates to the following via the above scheme around $v_2$ with radii $h$ and $2h$ respectively:
- $f(v_2-h) = f(v_2+h)$, i.e. $S_1 = S_3$
- $f(v_1-2h) = f(v_2 + 2h)$, i.e. $S_0 = S_4$

Plugging these into our original finite difference scheme for second derivative should give a boundary condition. 

Alternatively, we have a five point scheme for the derivative:
$$f'(x) \approx \frac{-f(x+2 h)+8 f(x+h)-8 f(x-h)+f(x-2h)}{12 h}$$

But, this may not remove the dependence on points outside the domain.
