```python
import numpy as np
import scipy.special
from scipy.integrate import trapezoid
from numpy import cos, sin
from numpy.linalg import norm
```

```python
def modmc1(m, q, x):
    return scipy.special.mathieu_modcem1(m, q, 180/np.pi*x)
def modmc2(m, q, x):
    return scipy.special.mathieu_modcem2(m, q, 180/np.pi*x)

def modms1(m, q, x):
    return scipy.special.mathieu_modsem1(m, q, 180/np.pi*x)
def modms2(m, q, x):
    return scipy.special.mathieu_modsem2(m, q, 180/np.pi*x)
```

```python
from scipy.linalg import det
import statistics

# m is (1 to 35)
# q cant support negative values on scipy 
# v is angle
# [0] is the value, [1] is the derivative

Q = np.array([ .001, .01, .1, 1, 10, 100])
M = range(21)
N = 200
V = np.linspace(6, 15, N)



for q in Q:
    for m in M:
        for v in V:
            
            [y1,y1d] = modmc1(m,q,v)
            [y2,y2d] = modmc2(m,q,v)

            w = y1 * y2d - y1d * y2
            wdiff = np.linalg.norm(w - (2 / np.pi)) / np.linalg.norm((2 / np.pi))
            
            if not np.isclose(wdiff, 0, atol=1e-6):
                print(f'q = {q}, m = {m}, v = {v}, diff = {wdiff}')
           

```
